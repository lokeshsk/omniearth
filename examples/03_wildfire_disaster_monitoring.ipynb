Combine **NASA FIRMS Active Fire Hotspots**, **Sentinel-5P TROPOMI Air Quality**, and **Sentinel-1 SAR** (cloud/smoke-penetrating radar) using zero credentials.

```python
from omniearth import OmniEarth

client = OmniEarth()
bbox = [-121.8, 39.6, -121.2, 40.1]  # Northern California sample AOI

fires = client.search(category="fire", bbox=bbox, limit=10)
atmosphere = client.search(category="atmosphere", bbox=bbox, limit=3)
sar = client.search(category="sar", bbox=bbox, limit=5)

print(f"Fire records: {len(fires)} | Atmosphere: {len(atmosphere)} | SAR scenes: {len(sar)}")
fires.to_geodataframe().head()
```
