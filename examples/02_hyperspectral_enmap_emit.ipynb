Query **224-band DLR EnMAP** and **285-band NASA EMIT** imaging spectroscopy archives through a single unified STAC interface.

```python
from omniearth import OmniEarth

client = OmniEarth()
aoi_bbox = [78.40, 20.55, 78.75, 20.85]

# Search NASA EMIT (285 spectral bands, 381-2493 nm)
emit_scenes = client.search(
    category="hyperspectral",
    provider="nasa_emit",
    bbox=aoi_bbox,
    limit=5,
)
print(f"NASA EMIT scenes found: {len(emit_scenes)}")

# Search DLR EnMAP (224 spectral bands, 420-2450 nm)
enmap_scenes = client.search(
    category="hyperspectral",
    provider="dlr_enmap",
    bbox=aoi_bbox,
    limit=5,
)
print(f"DLR EnMAP scenes found: {len(enmap_scenes)}")
if client.get_warnings():
    print("Provider health warnings:", client.get_warnings())
```

```python
# Export unified footprints for QGIS / Kepler.gl
if len(emit_scenes) > 0:
    emit_scenes.export_geojson("emit_footprints.geojson")
    gdf = emit_scenes.to_geodataframe()
    display(gdf[["id", "provider", "datetime", "geometry"]].head())
```
